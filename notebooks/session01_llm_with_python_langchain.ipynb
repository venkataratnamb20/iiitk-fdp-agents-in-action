## LLM Chatbot with memory/History

````python
import gradio as gr

messages = []

# python function- small change for gradio ui
def run_llm(
        message, history=[],
        model='gpt-4.1',
        temperature=0.7,
        max_tokens=1000
):
    if 'anthropic' in model:
        temperature = temperature if temperature < 1 else 1

    llm = init_chat_model(
        model=model,
        temperature=temperature,
        max_tokens=max_tokens
    )
    messages.append({'role': 'user', 'content': message})
    response = llm.invoke(messages)
    messages.append({'role': 'assistant', 'content': response.content})
    return response.content, response.response_metadata

with gr.Blocks(fill_width=True, fill_height=True) as app:
    with gr.Sidebar(label='Model Settings', position='right'):
        model_comp = gr.Dropdown(models, label='Model', value='openai:gpt-4.1')
        temperature = gr.Number(label='Temperature', value=0.7, minimum=0.0, maximum=2.0)
        max_tokens = gr.Number(label='Max Tokens', value=1000)

    with gr.Row(equal_height=True):
        model_info = gr.JSON(label='Model Info')
        # create chat inerface
        chat_ui = gr.ChatInterface(
            run_llm,
            additional_inputs=[
                model_comp,
                temperature,
                max_tokens
            ],
            additional_outputs=[
                model_info
            ],
        )

# launch ui
app.launch(theme='soft', debug=False)
````

## LLM Chatbot with Memory and System Prompt

````python
import gradio as gr

messages = []

# python function- small change for gradio ui
def run_llm(
        message, history=[],
        model='gpt-4.1',
        temperature=0.7,
        max_tokens=1000,
        system_prompt='You are a helpful assistant.'
):
    if 'anthropic' in model:
        temperature = temperature if temperature < 1 else 1

    llm = init_chat_model(
        model=model,
        temperature=temperature,
        max_tokens=max_tokens
    )

    # Add system prompt to history-messages
    # System message should be the first message in the history.
    if not messages:
        messages.append({'role': 'system', 'content': system_prompt})
    else:
        messages[0] = {'role': 'system', 'content': system_prompt}

    # add user message to memory- messages
    messages.append({'role': 'user', 'content': message})
    response = llm.invoke(messages)
    # add ai response to memory-messages
    messages.append({'role': 'assistant', 'content': response.content})
    return response.content, response.response_metadata

with gr.Blocks(fill_width=True, fill_height=True) as app:
    with gr.Sidebar(label='Model Settings', position='right'):
        model_comp = gr.Dropdown(models, label='Model', value='openai:gpt-4.1')
        temperature = gr.Number(label='Temperature', value=0.7, minimum=0.0, maximum=2.0)
        max_tokens = gr.Number(label='Max Tokens', value=1000)
        system_prompt = gr.Textbox(label='System Message')

    with gr.Row(equal_height=True):
        model_info = gr.JSON(label='Model Info')
        # create chat inerface
        chat_ui = gr.ChatInterface(
            run_llm,
            additional_inputs=[
                model_comp,
                temperature,
                max_tokens,
                system_prompt
            ],
            additional_outputs=[
                model_info
            ],
        )

# launch ui
app.launch(theme='soft', debug=False)
````

````python
# import packages
import os
from google.colab import userdata

from langchain.chat_models import init_chat_model

from pydantic import BaseModel, Field

# set environment variable
os.environ['OPENAI_API_KEY'] = userdata.get('OPENAI_API_KEY')

# python list as memory
messages = []

# create llm
llm = init_chat_model(
    model='gpt-4.1'
)

# llm with structured response
class Code(BaseModel):
    title: str = Field(description='Title of the software program')
    code: str = Field(description='Software program as per user requirement')
    language: str = Field(description='Programming language used to write the code')
    doc: str = Field(description='Documentation of the software program for developers and users in markdown forrmat')

# bind response format to the llm
llm_with_structured_format = llm.with_structured_output(Code)

# run llm
# response = llm.invoke('Who are you?')
# response = llm.invoke(['Who are you?'])

# user message
user_msg = {'role': 'user', 'content': 'Write python code for prime number'}
# add user message to memory
messages.append(user_msg)
# pass messages to llm
response = llm_with_structured_format.invoke(messages)

# add ai response to messages
# ai message
ai_response = f"""
# {response.title}

## Program

```{response.language.strip().lower()}
{response.code}
```

## Documentation

{response.doc}

"""
ai_msg = {'role': 'assistant', 'content': ai_response}
messages.append(ai_msg)


# print(response.code)
response
````

## LLM Chatbot with Structured Response


````python
import gradio as gr

# llm with structured response
class Code(BaseModel):
    title: str = Field(
        default='',
        description='Title of the software program'
    )
    code: str = Field(
        default='',
        description='Software program as per user requirement'
    )
    language: str = Field(
        default='python',
        description='Programming language used to write the code'
    )
    doc: str = Field(description='Documentation of the software program for developers and users in markdown forrmat or the general response')

messages = []

# python function- small change for gradio ui
def run_llm(
        message, history=[],
        model='gpt-4.1',
        temperature=0.7,
        max_tokens=1000,
        system_prompt='You are a helpful assistant.'
):
    if 'anthropic' in model:
        temperature = temperature if temperature < 1 else 1

    llm = init_chat_model(
        model=model,
        temperature=temperature,
        max_tokens=max_tokens
    )

    # bind response format to the llm
    llm_with_structured_format = llm.with_structured_output(Code)


    # Add system prompt to history-messages
    # System message should be the first message in the history.
    if not messages:
        messages.append({'role': 'system', 'content': system_prompt})
    else:
        messages[0] = {'role': 'system', 'content': system_prompt}

    # add user message to memory- messages
    messages.append({'role': 'user', 'content': message})
    response = llm_with_structured_format.invoke(messages)
    # add ai response to memory-messages
    # ai message
    ai_response = f"""
    # {response.title}

    ## Program

    ```{response.language.strip().lower()}
    {response.code.strip()}
    ```

    ## Documentation

    {response.doc}

    """
    messages.append({'role': 'assistant', 'content': ai_response})
    code_component = gr.Code(
        label = 'Code Editor',
        value=response.code,
        language=response.language.strip().lower(),
        show_line_numbers=True,
        autocomplete=True,
        interactive=True,
    )
    return response.doc, code_component

with gr.Blocks(fill_width=True, fill_height=True) as app:
    with gr.Sidebar(label='Model Settings', position='right'):
        model_comp = gr.Dropdown(models, label='Model', value='openai:gpt-4.1')
        temperature = gr.Number(label='Temperature', value=0.7, minimum=0.0, maximum=2.0)
        max_tokens = gr.Number(label='Max Tokens', value=1000)
        system_prompt = gr.Textbox(label='System Message')

    with gr.Row(equal_height=True):
        code_editor = gr.Code(label='Code Editor')
        # create chat inerface
        chat_ui = gr.ChatInterface(
            run_llm,
            additional_inputs=[
                model_comp,
                temperature,
                max_tokens,
                system_prompt
            ],
            additional_outputs=[
                code_editor
            ],
        )

# launch ui
app.launch(theme='soft', debug=False)
````
